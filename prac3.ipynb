```python
rules = [
    (["fever", "cough"], "flu"),
    (["flu", "body pain"], "see_doctor"),
    (["sneezing", "running nose"], "cold"),
    (["cold"], "rest"),
]

facts = ["fever", "cough", "body pain"]
print("Rules :")
for r in rules:
    print("  IF", " AND ".join(r[0]), "Then", r[1])
print("facts", facts)
```

```python
def forward_chaining(rules, facts, goal):
    known = list(facts)
    changed = True
    while changed:
        changed = False
        for condition, conclusion in rules:
            all_true = True
            for c in condition:
                if c not in known:
                    all_true = False
            if all_true and conclusion not in known:
                known.append(conclusion)
                print("rule fired :", condition, "->", conclusion)
                changed = True
    return goal in known
```

```python
print("Goals reached ?", forward_chaining(rules, facts, "see_doctor"))
```

```python
def backward_chaining(rules, facts, goal):
    if goal in facts:
        print("Fact found:", goal)
        return True

    for condition, conclusion in rules:
        if conclusion == goal:
            print("To prove ", goal, "We need", condition)
            all_proved = True
            for c in condition:
                if not backward_chaining(rules, facts, c):
                    all_proved = False
            if all_proved:
                return True
    return False
```

```python
print("Goal reached ?", backward_chaining(rules, facts, "see_doctor"))
```

```python
print("Forward chaining  :", forward_chaining(rules, facts, "rest"))
print("Backward chaining :", backward_chaining(rules, facts, "rest"))
```
