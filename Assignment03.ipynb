```python
import tkinter as tk
from tkinter import ttk, messagebox
import random
import time
from collections import deque
```

```python
# ============================================================
# BFS ALGORITHM
# ============================================================

def bfs(maze, start, goal):

    rows = len(maze)
    cols = len(maze[0])

    queue = deque([start])

    visited = {start}

    parent = {}

    visited_order = []

    operations = 0

    directions = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    while queue:

        current = queue.popleft()

        operations += 1

        visited_order.append(current)

        if current == goal:
            break

        r, c = current

        for dr, dc in directions:

            nr = r + dr
            nc = c + dc

            # Boundary check
            if nr < 0 or nr >= rows:
                continue

            if nc < 0 or nc >= cols:
                continue

            # Stoppage
            if maze[nr][nc] == 1:
                continue

            neighbor = (nr, nc)

            # Already visited
            if neighbor in visited:
                continue

            visited.add(neighbor)

            parent[neighbor] = current

            queue.append(neighbor)

    # --------------------------------------------------------
    # Construct path
    # --------------------------------------------------------

    path = []

    if goal in visited:

        current = goal

        while current != start:

            path.append(current)

            current = parent[current]

        path.append(start)

        path.reverse()

    return path, visited_order, operations


# ============================================================
# DFS ALGORITHM
# ============================================================

def dfs(maze, start, goal):

    rows = len(maze)
    cols = len(maze[0])

    stack = [start]

    visited = {start}

    parent = {}

    visited_order = []

    operations = 0

    directions = [
        (-1, 0),   # Up
        (1, 0),    # Down
        (0, -1),   # Left
        (0, 1)     # Right
    ]

    while stack:

        current = stack.pop()

        operations += 1

        visited_order.append(current)

        if current == goal:
            break

        r, c = current

        for dr, dc in directions:

            nr = r + dr
            nc = c + dc

            # Boundary check
            if nr < 0 or nr >= rows:
                continue

            if nc < 0 or nc >= cols:
                continue

            # Stoppage
            if maze[nr][nc] == 1:
                continue

            neighbor = (nr, nc)

            # Already visited
            if neighbor in visited:
                continue

            visited.add(neighbor)

            parent[neighbor] = current

            stack.append(neighbor)

    # --------------------------------------------------------
    # Construct path
    # --------------------------------------------------------

    path = []

    if goal in visited:

        current = goal

        while current != start:

            path.append(current)

            current = parent[current]

        path.append(start)

        path.reverse()

    return path, visited_order, operations
```

```python
# ============================================================
# BFS vs DFS MAZE SOLVER GUI
# ============================================================

class MazeSolverGUI:

    def __init__(self, root):

        # ====================================================
        # ROOT WINDOW
        # ====================================================

        self.root = root

        self.root.title(
            "BFS vs DFS Maze Solver"
        )

        # Start maximized
        try:
            self.root.state("zoomed")
        except:
            self.root.geometry("1400x900")

        # Allow resizing
        self.root.resizable(True, True)

        # ====================================================
        # VARIABLES
        # ====================================================

        self.maze = None

        self.start = None

        self.goal = None

        self.current_path = []

        self.visited_order = []

        self.current_step = 0

        self.running = False

        self.cell_size = 30

        self.offset_x = 0

        self.offset_y = 0

        self.final_operations = 0

        self.final_time = 0

        self.final_algorithm = ""

        # ====================================================
        # COLORS
        # ====================================================

        self.COLOR_FREE = "white"

        self.COLOR_WALL = "black"

        self.COLOR_START = "green"

        self.COLOR_GOAL = "red"

        self.COLOR_VISITED = "lightblue"

        self.COLOR_PATH = "yellow"

        # ====================================================
        # TITLE
        # ====================================================

        title = tk.Label(
            self.root,
            text="BFS vs DFS Maze Solver",
            font=("Arial", 26, "bold")
        )

        title.pack(
            pady=15
        )

        # ====================================================
        # CONTROL PANEL
        # ====================================================

        control_frame = tk.Frame(
            self.root
        )

        control_frame.pack(
            pady=5
        )

        # ----------------------------------------------------
        # Matrix Size
        # ----------------------------------------------------

        tk.Label(
            control_frame,
            text="Matrix Size:",
            font=("Arial", 11)
        ).grid(
            row=0,
            column=0,
            padx=5,
            pady=5
        )

        self.size_entry = tk.Entry(
            control_frame,
            width=8,
            font=("Arial", 11)
        )

        self.size_entry.insert(
            0,
            "15"
        )

        self.size_entry.grid(
            row=0,
            column=1,
            padx=5
        )

        # ----------------------------------------------------
        # Obstacles
        # ----------------------------------------------------

        tk.Label(
            control_frame,
            text="Obstacles %:",
            font=("Arial", 11)
        ).grid(
            row=0,
            column=2,
            padx=5
        )

        self.obstacle_entry = tk.Entry(
            control_frame,
            width=8,
            font=("Arial", 11)
        )

        self.obstacle_entry.insert(
            0,
            "25"
        )

        self.obstacle_entry.grid(
            row=0,
            column=3,
            padx=5
        )

        # ----------------------------------------------------
        # Generate Maze
        # ----------------------------------------------------

        self.generate_button = tk.Button(
            control_frame,
            text="Generate Maze",
            command=self.generate_maze,
            bg="lightblue",
            font=("Arial", 11, "bold"),
            width=16
        )

        self.generate_button.grid(
            row=0,
            column=4,
            padx=15
        )

        # ----------------------------------------------------
        # Algorithm
        # ----------------------------------------------------

        tk.Label(
            control_frame,
            text="Algorithm:",
            font=("Arial", 11)
        ).grid(
            row=1,
            column=0,
            padx=5,
            pady=10
        )

        self.algorithm = ttk.Combobox(
            control_frame,
            values=[
                "BFS",
                "DFS"
            ],
            state="readonly",
            width=12,
            font=("Arial", 11)
        )

        self.algorithm.set(
            "BFS"
        )

        self.algorithm.grid(
            row=1,
            column=1
        )

        # ----------------------------------------------------
        # Animation Speed
        # ----------------------------------------------------

        tk.Label(
            control_frame,
            text="Speed (ms):",
            font=("Arial", 11)
        ).grid(
            row=1,
            column=2
        )

        self.speed_entry = tk.Entry(
            control_frame,
            width=8,
            font=("Arial", 11)
        )

        self.speed_entry.insert(
            0,
            "40"
        )

        self.speed_entry.grid(
            row=1,
            column=3
        )

        # ----------------------------------------------------
        # Run Algorithm
        # ----------------------------------------------------

        self.run_button = tk.Button(
            control_frame,
            text="Run Algorithm",
            command=self.run_algorithm,
            bg="lightgreen",
            font=("Arial", 11, "bold"),
            width=16
        )

        self.run_button.grid(
            row=1,
            column=4,
            padx=15
        )

        # ----------------------------------------------------
        # Compare BFS vs DFS
        # ----------------------------------------------------

        self.compare_button = tk.Button(
            control_frame,
            text="Compare BFS vs DFS",
            command=self.compare_algorithms,
            bg="orange",
            font=("Arial", 11, "bold"),
            width=20
        )

        self.compare_button.grid(
            row=2,
            column=1,
            columnspan=3,
            pady=10
        )

        # ====================================================
        # MAZE CANVAS
        # ====================================================

        self.canvas = tk.Canvas(
            self.root,
            bg="white",
            highlightthickness=2,
            highlightbackground="black"
        )

        self.canvas.pack(
            fill="both",
            expand=True,
            padx=25,
            pady=10
        )

        # ====================================================
        # LEGEND
        # ====================================================

        legend = tk.Label(
            self.root,
            text=(
                "Green = Start     "
                "Red = Goal     "
                "Black = Stoppage     "
                "Blue = Explored     "
                "Yellow = Final Path"
            ),
            font=("Arial", 11)
        )

        legend.pack(
            pady=5
        )

        # ====================================================
        # RESULT PANEL
        # ====================================================

        self.result_text = tk.Text(
            self.root,
            height=9,
            width=110,
            font=("Courier New", 11)
        )

        self.result_text.pack(
            pady=10
        )

        # ====================================================
        # WINDOW RESIZE EVENT
        # ====================================================

        self.root.bind(
            "<Configure>",
            self.window_resized
        )

        # ====================================================
        # GENERATE FIRST MAZE
        # ====================================================

        self.root.after(
            200,
            self.generate_maze
        )

    # ========================================================
    # WINDOW RESIZED
    # ========================================================

    def window_resized(self, event):

        if self.maze is None:
            return

        # Do not redraw while animation is running
        if self.running:
            return

        self.draw_maze()

    # ========================================================
    # GENERATE MAZE
    # ========================================================

    def generate_maze(self):

        if self.running:
            return

        # ----------------------------------------------------
        # Read input
        # ----------------------------------------------------

        try:

            size = int(
                self.size_entry.get()
            )

            obstacle_percentage = float(
                self.obstacle_entry.get()
            )

        except ValueError:

            messagebox.showerror(
                "Invalid Input",
                "Please enter valid numbers."
            )

            return

        # ----------------------------------------------------
        # Validate size
        # ----------------------------------------------------

        if size < 5 or size > 40:

            messagebox.showerror(
                "Invalid Matrix Size",
                "Matrix size must be between 5 and 40."
            )

            return

        # ----------------------------------------------------
        # Validate obstacle percentage
        # ----------------------------------------------------

        if (
            obstacle_percentage < 0
            or
            obstacle_percentage > 45
        ):

            messagebox.showerror(
                "Invalid Obstacles",
                "Obstacle percentage must be between 0 and 45."
            )

            return

        # ----------------------------------------------------
        # Create matrix
        # ----------------------------------------------------

        self.maze = [
            [0 for _ in range(size)]
            for _ in range(size)
        ]

        # ----------------------------------------------------
        # Random obstacles
        # ----------------------------------------------------

        probability = (
            obstacle_percentage / 100
        )

        for i in range(size):

            for j in range(size):

                if random.random() < probability:

                    self.maze[i][j] = 1

        # ----------------------------------------------------
        # Start and Goal
        # ----------------------------------------------------

        self.start = (
            0,
            0
        )

        self.goal = (
            size - 1,
            size - 1
        )

        # Ensure start is free

        self.maze[
            self.start[0]
        ][
            self.start[1]
        ] = 0

        # Ensure goal is free

        self.maze[
            self.goal[0]
        ][
            self.goal[1]
        ] = 0

        # ----------------------------------------------------
        # Reset
        # ----------------------------------------------------

        self.current_path = []

        self.visited_order = []

        self.current_step = 0

        self.running = False

        # ----------------------------------------------------
        # Draw
        # ----------------------------------------------------

        self.draw_maze()

        # ----------------------------------------------------
        # Results
        # ----------------------------------------------------

        self.result_text.delete(
            "1.0",
            tk.END
        )

        self.result_text.insert(
            tk.END,
            "NEW MAZE GENERATED\n"
        )

        self.result_text.insert(
            tk.END,
            "-" * 60 + "\n"
        )

        self.result_text.insert(
            tk.END,
            f"Matrix Size : {size} x {size}\n"
        )

        self.result_text.insert(
            tk.END,
            f"Obstacles   : {obstacle_percentage}%\n"
        )

        self.result_text.insert(
            tk.END,
            f"Start       : {self.start}\n"
        )

        self.result_text.insert(
            tk.END,
            f"Goal        : {self.goal}\n"
        )

    # ========================================================
    # DRAW MAZE
    # ========================================================

    def draw_maze(self):

        self.canvas.delete(
            "all"
        )

        if self.maze is None:
            return

        size = len(
            self.maze
        )

        # ----------------------------------------------------
        # Get actual canvas size
        # ----------------------------------------------------

        canvas_width = (
            self.canvas.winfo_width()
        )

        canvas_height = (
            self.canvas.winfo_height()
        )

        # Fallback
        if canvas_width <= 1:

            canvas_width = 800

        if canvas_height <= 1:

            canvas_height = 500

        # ----------------------------------------------------
        # Calculate cell size
        # ----------------------------------------------------

        self.cell_size = min(
            canvas_width // size,
            canvas_height // size
        )

        # Prevent extremely small cells

        self.cell_size = max(
            self.cell_size,
            5
        )

        # ----------------------------------------------------
        # Calculate maze size
        # ----------------------------------------------------

        maze_width = (
            size *
            self.cell_size
        )

        maze_height = (
            size *
            self.cell_size
        )

        # ----------------------------------------------------
        # Center maze
        # ----------------------------------------------------

        self.offset_x = (
            canvas_width -
            maze_width
        ) // 2

        self.offset_y = (
            canvas_height -
            maze_height
        ) // 2

        # ----------------------------------------------------
        # Draw every cell
        # ----------------------------------------------------

        for i in range(size):

            for j in range(size):

                x1 = (
                    self.offset_x +
                    j * self.cell_size
                )

                y1 = (
                    self.offset_y +
                    i * self.cell_size
                )

                x2 = (
                    x1 +
                    self.cell_size
                )

                y2 = (
                    y1 +
                    self.cell_size
                )

                # Free cell

                color = (
                    self.COLOR_FREE
                )

                # Stoppage

                if self.maze[i][j] == 1:

                    color = (
                        self.COLOR_WALL
                    )

                # Start

                if (i, j) == self.start:

                    color = (
                        self.COLOR_START
                    )

                # Goal

                if (i, j) == self.goal:

                    color = (
                        self.COLOR_GOAL
                    )

                self.canvas.create_rectangle(
                    x1,
                    y1,
                    x2,
                    y2,
                    fill=color,
                    outline="gray"
                )

    # ========================================================
    # DRAW CELL
    # ========================================================

    def draw_cell(
        self,
        cell,
        color
    ):

        r, c = cell

        x1 = (
            self.offset_x +
            c * self.cell_size
        )

        y1 = (
            self.offset_y +
            r * self.cell_size
        )

        x2 = (
            x1 +
            self.cell_size
        )

        y2 = (
            y1 +
            self.cell_size
        )

        self.canvas.create_rectangle(
            x1,
            y1,
            x2,
            y2,
            fill=color,
            outline="gray"
        )

    # ========================================================
    # RUN ALGORITHM
    # ========================================================

    def run_algorithm(self):

        if self.maze is None:
            return

        if self.running:
            return

        algorithm = (
            self.algorithm.get()
        )

        # ----------------------------------------------------
        # Read animation speed
        # ----------------------------------------------------

        try:

            delay = int(
                self.speed_entry.get()
            )

        except ValueError:

            messagebox.showerror(
                "Invalid Speed",
                "Enter speed in milliseconds."
            )

            return

        if delay < 1:

            delay = 1

        # ----------------------------------------------------
        # Measure algorithm execution
        # ----------------------------------------------------

        start_time = (
            time.perf_counter()
        )

        if algorithm == "BFS":

            path, visited, operations = bfs(
                self.maze,
                self.start,
                self.goal
            )

        else:

            path, visited, operations = dfs(
                self.maze,
                self.start,
                self.goal
            )

        execution_time = (
            time.perf_counter()
            -
            start_time
        )

        # ----------------------------------------------------
        # Store results
        # ----------------------------------------------------

        self.current_path = path

        self.visited_order = visited

        self.current_step = 0

        self.final_operations = operations

        self.final_time = execution_time

        self.final_algorithm = algorithm

        self.running = True

        # ----------------------------------------------------
        # Reset maze
        # ----------------------------------------------------

        self.draw_maze()

        # ----------------------------------------------------
        # Start animation
        # ----------------------------------------------------

        self.animate_search(
            delay
        )

    # ========================================================
    # ANIMATE SEARCH
    # ========================================================

    def animate_search(
        self,
        delay
    ):

        if (
            self.current_step
            <
            len(self.visited_order)
        ):

            cell = (
                self.visited_order[
                    self.current_step
                ]
            )

            # Don't overwrite Start / Goal

            if (
                cell != self.start
                and
                cell != self.goal
            ):

                self.draw_cell(
                    cell,
                    self.COLOR_VISITED
                )

            self.current_step += 1

            # Update screen

            self.root.update_idletasks()

            # Schedule next step

            self.root.after(
                delay,
                lambda:
                self.animate_search(
                    delay
                )
            )

            return

        # Search finished

        self.animate_path(
            0,
            delay
        )

    # ========================================================
    # ANIMATE FINAL PATH
    # ========================================================

    def animate_path(
        self,
        index,
        delay
    ):

        if (
            index
            <
            len(self.current_path)
        ):

            cell = (
                self.current_path[
                    index
                ]
            )

            # Don't overwrite Start / Goal

            if (
                cell != self.start
                and
                cell != self.goal
            ):

                self.draw_cell(
                    cell,
                    self.COLOR_PATH
                )

            self.root.update_idletasks()

            self.root.after(
                delay,
                lambda:
                self.animate_path(
                    index + 1,
                    delay
                )
            )

            return

        # Animation complete

        self.running = False

        self.show_results()

    # ========================================================
    # SHOW RESULTS
    # ========================================================

    def show_results(self):

        self.result_text.delete(
            "1.0",
            tk.END
        )

        self.result_text.insert(
            tk.END,
            "=" * 70 + "\n"
        )

        self.result_text.insert(
            tk.END,
            f"{self.final_algorithm} RESULTS\n"
        )

        self.result_text.insert(
            tk.END,
            "=" * 70 + "\n\n"
        )

        # ----------------------------------------------------
        # Path
        # ----------------------------------------------------

        if self.current_path:

            path_length = (
                len(self.current_path) - 1
            )

            self.result_text.insert(
                tk.END,
                f"Path Length     : "
                f"{path_length} moves\n"
            )

        else:

            self.result_text.insert(
                tk.END,
                "Path Length     : NO PATH\n"
            )

        # ----------------------------------------------------
        # Nodes
        # ----------------------------------------------------

        self.result_text.insert(
            tk.END,
            f"Nodes Explored  : "
            f"{len(self.visited_order)}\n"
        )

        # ----------------------------------------------------
        # Operations
        # ----------------------------------------------------

        self.result_text.insert(
            tk.END,
            f"Operations      : "
            f"{self.final_operations}\n"
        )

        # ----------------------------------------------------
        # Execution Time
        # ----------------------------------------------------

        self.result_text.insert(
            tk.END,
            f"Execution Time  : "
            f"{self.final_time:.8f} seconds\n"
        )

        # ----------------------------------------------------
        # Shortest Path
        # ----------------------------------------------------

        if self.final_algorithm == "BFS":

            self.result_text.insert(
                tk.END,
                "\nShortest Path   : YES\n"
            )

        else:

            self.result_text.insert(
                tk.END,
                "\nShortest Path   : NOT GUARANTEED\n"
            )

    # ========================================================
    # COMPARE BFS VS DFS
    # ========================================================

    def compare_algorithms(self):

        if self.maze is None:

            return

        if self.running:

            return

        # ----------------------------------------------------
        # BFS
        # ----------------------------------------------------

        start_time = (
            time.perf_counter()
        )

        bfs_path, bfs_visited, bfs_operations = bfs(
            self.maze,
            self.start,
            self.goal
        )

        bfs_time = (
            time.perf_counter()
            -
            start_time
        )

        # ----------------------------------------------------
        # DFS
        # ----------------------------------------------------

        start_time = (
            time.perf_counter()
        )

        dfs_path, dfs_visited, dfs_operations = dfs(
            self.maze,
            self.start,
            self.goal
        )

        dfs_time = (
            time.perf_counter()
            -
            start_time
        )

        # ----------------------------------------------------
        # Path lengths
        # ----------------------------------------------------

        if bfs_path:

            bfs_length = (
                len(bfs_path) - 1
            )

        else:

            bfs_length = "NO PATH"

        if dfs_path:

            dfs_length = (
                len(dfs_path) - 1
            )

        else:

            dfs_length = "NO PATH"

        # ----------------------------------------------------
        # Display comparison
        # ----------------------------------------------------

        self.result_text.delete(
            "1.0",
            tk.END
        )

        self.result_text.insert(
            tk.END,
            "=" * 80 + "\n"
        )

        self.result_text.insert(
            tk.END,
            "BFS vs DFS PERFORMANCE COMPARISON\n"
        )

        self.result_text.insert(
            tk.END,
            "=" * 80 + "\n\n"
        )

        self.result_text.insert(
            tk.END,
            f"{'Metric':<25}"
            f"{'BFS':<25}"
            f"{'DFS':<25}\n"
        )

        self.result_text.insert(
            tk.END,
            "-" * 75 + "\n"
        )

        self.result_text.insert(
            tk.END,
            f"{'Path Length':<25}"
            f"{str(bfs_length):<25}"
            f"{str(dfs_length):<25}\n"
        )

        self.result_text.insert(
            tk.END,
            f"{'Nodes Explored':<25}"
            f"{len(bfs_visited):<25}"
            f"{len(dfs_visited):<25}\n"
        )

        self.result_text.insert(
            tk.END,
            f"{'Operations':<25}"
            f"{bfs_operations:<25}"
            f"{dfs_operations:<25}\n"
        )

        self.result_text.insert(
            tk.END,
            f"{'Execution Time':<25}"
            f"{bfs_time:<25.8f}"
            f"{dfs_time:<25.8f}\n"
        )

        self.result_text.insert(
            tk.END,
            "\n"
        )

        self.result_text.insert(
            tk.END,
            "CONCLUSION\n"
        )

        self.result_text.insert(
            tk.END,
            "-" * 75 + "\n"
        )

        self.result_text.insert(
            tk.END,
            "BFS guarantees the shortest path in an "
            "unweighted maze.\n"
        )

        self.result_text.insert(
            tk.END,
            "DFS does not guarantee the shortest path.\n"
        )
```

```python
# Close any previous Tkinter window if necessary
try:
    root.destroy()
except:
    pass

# Create new application
root = tk.Tk()

app = MazeSolverGUI(root)

root.mainloop()
```
