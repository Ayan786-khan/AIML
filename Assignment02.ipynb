```python
import tkinter as tk
from tkinter import ttk
```

```python
# ============================================================
# ENVIRONMENT
# ============================================================

class Environment:

    def __init__(self):

        # ----------------------------------------------------
        # ENVIRONMENT STATE
        # ----------------------------------------------------

        self.total_spaces = 20
        self.available_spaces = 10
        self.cars_waiting = 2
        self.reserved_available = True

        self.gate_open = False
        self.last_action = "NONE"

        # Agents will be connected later
        self.simple_agent = None
        self.model_agent = None

        # ----------------------------------------------------
        # CREATE MAIN WINDOW
        # ----------------------------------------------------

        self.root = tk.Tk()

        self.root.title(
            "Intelligent Smart Parking Agent"
        )

        self.root.geometry(
            "800x700"
        )

        self.root.minsize(
            650,
            500
        )

        # ----------------------------------------------------
        # CREATE SCROLLABLE CENTERED AREA
        # ----------------------------------------------------

        self.canvas = tk.Canvas(
            self.root,
            highlightthickness=0
        )

        self.scrollbar = ttk.Scrollbar(
            self.root,
            orient="vertical",
            command=self.canvas.yview
        )

        self.scrollable_frame = tk.Frame(
            self.canvas
        )

        # Create window inside canvas
        self.canvas_window = self.canvas.create_window(
            (0, 0),
            window=self.scrollable_frame,
            anchor="n"
        )

        # Update scroll region
        self.scrollable_frame.bind(
            "<Configure>",
            lambda event: self.canvas.configure(
                scrollregion=self.canvas.bbox("all")
            )
        )

        # Make frame width equal to canvas width
        self.canvas.bind(
            "<Configure>",
            self.resize_canvas_frame
        )

        self.canvas.configure(
            yscrollcommand=self.scrollbar.set
        )

        self.canvas.pack(
            side="left",
            fill="both",
            expand=True
        )

        self.scrollbar.pack(
            side="right",
            fill="y"
        )

        # ----------------------------------------------------
        # CREATE GUI
        # ----------------------------------------------------

        self.create_gui()

    # ========================================================
    # CENTER SCROLLABLE FRAME
    # ========================================================

    def resize_canvas_frame(self, event):

        self.canvas.itemconfig(
            self.canvas_window,
            width=event.width
        )

    # ========================================================
    # CREATE GUI
    # ========================================================

    def create_gui(self):

        # ----------------------------------------------------
        # MAIN CONTENT FRAME
        # ----------------------------------------------------

        self.main_frame = tk.Frame(
            self.scrollable_frame,
            padx=30
        )

        self.main_frame.pack(
            fill="x"
        )

        # ----------------------------------------------------
        # TITLE
        # ----------------------------------------------------

        title = tk.Label(
            self.main_frame,
            text="Smart Parking Intelligent Agent",
            font=("Arial", 22, "bold")
        )

        title.pack(
            pady=20
        )

        subtitle = tk.Label(
            self.main_frame,
            text="PEAS + Simple Reflex + Model-Based Agent",
            font=("Arial", 11)
        )

        subtitle.pack(
            pady=(0, 15)
        )

        # ====================================================
        # INPUT FRAME
        # ====================================================

        input_frame = tk.LabelFrame(
            self.main_frame,
            text="Parking Environment Input",
            font=("Arial", 13, "bold"),
            padx=25,
            pady=20
        )

        input_frame.pack(
            pady=10
        )

        # ----------------------------------------------------
        # AVAILABLE SPACES
        # ----------------------------------------------------

        tk.Label(
            input_frame,
            text="Available Spaces:",
            font=("Arial", 12)
        ).grid(
            row=0,
            column=0,
            padx=15,
            pady=10,
            sticky="e"
        )

        self.available_entry = tk.Entry(
            input_frame,
            font=("Arial", 12),
            width=15
        )

        self.available_entry.grid(
            row=0,
            column=1,
            padx=15,
            pady=10
        )

        self.available_entry.insert(
            0,
            "10"
        )

        # ----------------------------------------------------
        # CARS WAITING
        # ----------------------------------------------------

        tk.Label(
            input_frame,
            text="Cars Waiting:",
            font=("Arial", 12)
        ).grid(
            row=1,
            column=0,
            padx=15,
            pady=10,
            sticky="e"
        )

        self.waiting_entry = tk.Entry(
            input_frame,
            font=("Arial", 12),
            width=15
        )

        self.waiting_entry.grid(
            row=1,
            column=1,
            padx=15,
            pady=10
        )

        self.waiting_entry.insert(
            0,
            "2"
        )

        # ----------------------------------------------------
        # RESERVED SLOT
        # ----------------------------------------------------

        tk.Label(
            input_frame,
            text="Reserved Slot Available:",
            font=("Arial", 12)
        ).grid(
            row=2,
            column=0,
            padx=15,
            pady=10,
            sticky="e"
        )

        self.reserved_var = tk.StringVar(
            value="Yes"
        )

        self.reserved_dropdown = ttk.Combobox(
            input_frame,
            textvariable=self.reserved_var,
            values=["Yes", "No"],
            state="readonly",
            width=13
        )

        self.reserved_dropdown.grid(
            row=2,
            column=1,
            padx=15,
            pady=10
        )

        # ====================================================
        # RUN BUTTON
        # ====================================================

        self.run_button = tk.Button(
            self.main_frame,
            text="RUN AGENTS",
            font=("Arial", 14, "bold"),
            command=self.run_agents,
            padx=35,
            pady=10
        )

        self.run_button.pack(
            pady=15
        )

        # ====================================================
        # ENVIRONMENT DISPLAY
        # ====================================================

        self.environment_frame = tk.LabelFrame(
            self.main_frame,
            text="Parking Lot Environment",
            font=("Arial", 13, "bold"),
            padx=25,
            pady=15
        )

        self.environment_frame.pack(
            fill="x",
            pady=10
        )

        self.environment_label = tk.Label(
            self.environment_frame,
            text="",
            font=("Arial", 12),
            justify="left"
        )

        self.environment_label.pack()

        # ====================================================
        # SIMPLE REFLEX AGENT
        # ====================================================

        self.simple_frame = tk.LabelFrame(
            self.main_frame,
            text="Simple Reflex Agent",
            font=("Arial", 13, "bold"),
            padx=25,
            pady=15
        )

        self.simple_frame.pack(
            fill="x",
            pady=10
        )

        self.simple_label = tk.Label(
            self.simple_frame,
            text="Waiting for input...",
            font=("Arial", 12),
            justify="left"
        )

        self.simple_label.pack()

        # ====================================================
        # MODEL-BASED AGENT
        # ====================================================

        self.model_frame = tk.LabelFrame(
            self.main_frame,
            text="Model-Based Agent",
            font=("Arial", 13, "bold"),
            padx=25,
            pady=15
        )

        self.model_frame.pack(
            fill="x",
            pady=10
        )

        self.model_label = tk.Label(
            self.model_frame,
            text="Waiting for input...",
            font=("Arial", 12),
            justify="left"
        )

        self.model_label.pack()

        # ====================================================
        # PEAS INFORMATION
        # ====================================================

        self.peas_frame = tk.LabelFrame(
            self.main_frame,
            text="PEAS Representation",
            font=("Arial", 13, "bold"),
            padx=25,
            pady=15
        )

        self.peas_frame.pack(
            fill="x",
            pady=10
        )

        peas_text = (
            "Performance Measure:\n"
            "Efficient parking allocation and reduced waiting time.\n\n"

            "Environment:\n"
            "Parking lot with parking spaces, cars and gate.\n\n"

            "Actuators:\n"
            "Parking gate and parking instructions.\n\n"

            "Sensors:\n"
            "Available-space sensor, vehicle sensor and reserved-space sensor."
        )

        self.peas_label = tk.Label(
            self.peas_frame,
            text=peas_text,
            font=("Arial", 11),
            justify="left"
        )

        self.peas_label.pack()

        # ----------------------------------------------------
        # INITIAL DISPLAY
        # ----------------------------------------------------

        self.update_gui()

    # ========================================================
    # GET USER INPUT
    # ========================================================

    def update_from_user(self):

        self.available_spaces = int(
            self.available_entry.get()
        )

        self.cars_waiting = int(
            self.waiting_entry.get()
        )

        self.reserved_available = (
            self.reserved_var.get() == "Yes"
        )

    # ========================================================
    # UPDATE ENVIRONMENT DISPLAY
    # ========================================================

    def update_gui(self):

        occupied = (
            self.total_spaces
            - self.available_spaces
        )

        self.environment_label.config(
            text=(
                f"Total Spaces: "
                f"{self.total_spaces}\n"

                f"Available Spaces: "
                f"{self.available_spaces}\n"

                f"Occupied Spaces: "
                f"{occupied}\n"

                f"Cars Waiting: "
                f"{self.cars_waiting}\n"

                f"Reserved Slot Available: "
                f"{self.reserved_available}\n"

                f"Gate: "
                f"{'OPEN' if self.gate_open else 'CLOSED'}\n"

                f"Last Action: "
                f"{self.last_action}"
            )
        )

    # ========================================================
    # RUN AGENTS
    # ========================================================

    def run_agents(self):

        try:

            # ------------------------------------------------
            # GET USER INPUT
            # ------------------------------------------------

            self.update_from_user()

            # ------------------------------------------------
            # SIMPLE REFLEX AGENT
            # ------------------------------------------------

            simple_action, simple_result = (
                self.simple_agent.run(self)
            )

            # ------------------------------------------------
            # MODEL-BASED AGENT
            # ------------------------------------------------

            model_data = (
                self.model_agent.run(self)
            )

            # =================================================
            # SIMPLE REFLEX OUTPUT
            # =================================================

            self.simple_label.config(
                text=(
                    f"Current Available Spaces: "
                    f"{self.available_spaces}\n"

                    f"Cars Waiting: "
                    f"{self.cars_waiting}\n\n"

                    f"Decision: "
                    f"{simple_action}\n"

                    f"Actuator: "
                    f"{simple_result}"
                )
            )

            # =================================================
            # MODEL-BASED DATA
            # =================================================

            previous_spaces = (
                model_data[
                    "previous_spaces"
                ]
            )

            trend = (
                model_data[
                    "availability_trend"
                ]
            )

            previous_action = (
                model_data[
                    "previous_action"
                ]
            )

            # ------------------------------------------------
            # FIRST RUN
            # ------------------------------------------------

            if previous_spaces is None:

                previous_text = "N/A"
                trend_text = "N/A"

            else:

                previous_text = str(
                    previous_spaces
                )

                trend_text = (
                    f"{trend:+d}"
                )

            # =================================================
            # MODEL-BASED OUTPUT
            # =================================================

            self.model_label.config(
                text=(
                    f"Current Available Spaces: "
                    f"{self.available_spaces}\n"

                    f"Cars Waiting: "
                    f"{self.cars_waiting}\n\n"

                    f"Decision: "
                    f"{model_data['action']}\n"

                    f"Actuator: "
                    f"{model_data['result']}\n\n"

                    f"INTERNAL STATE\n"
                    f"--------------------------\n"

                    f"Previous Spaces: "
                    f"{previous_text}\n"

                    f"Availability Trend: "
                    f"{trend_text}\n"

                    f"Previous Action: "
                    f"{previous_action}"
                )
            )

            # ------------------------------------------------
            # UPDATE ENVIRONMENT
            # ------------------------------------------------

            self.update_gui()

        except ValueError:

            self.environment_label.config(
                text=(
                    "ERROR\n\n"
                    "Please enter valid numbers "
                    "for Available Spaces and Cars Waiting."
                )
            )


# ============================================================
# SENSORS
# ============================================================

class Sensors:

    def read(self, environment):

        return {
            "available_spaces":
                environment.available_spaces,

            "cars_waiting":
                environment.cars_waiting,

            "reserved_available":
                environment.reserved_available
        }


# ============================================================
# ACTUATORS
# ============================================================

class Actuators:

    def execute(self, action, environment):

        # ----------------------------------------------------
        # ALLOW PARKING
        # ----------------------------------------------------

        if action == "ALLOW_PARKING":

            if environment.available_spaces > 0:

                environment.available_spaces -= 1

                environment.gate_open = True

                environment.last_action = (
                    "ALLOW_PARKING"
                )

                return (
                    "Gate opened - "
                    "Car allowed to park"
                )

            else:

                environment.gate_open = False

                environment.last_action = (
                    "NO_PARKING"
                )

                return (
                    "Parking full - "
                    "Gate remains closed"
                )

        # ----------------------------------------------------
        # WAIT
        # ----------------------------------------------------

        elif action == "WAIT":

            environment.gate_open = False

            environment.last_action = "WAIT"

            return (
                "Gate closed - "
                "Car must wait"
            )

        # ----------------------------------------------------
        # NO PARKING
        # ----------------------------------------------------

        elif action == "NO_PARKING":

            environment.gate_open = False

            environment.last_action = (
                "NO_PARKING"
            )

            return (
                "No parking available - "
                "Gate closed"
            )

        return "Unknown action"


# ============================================================
# SIMPLE REFLEX AGENT
# ============================================================

class SimpleReflexAgent:

    def __init__(self):

        self.sensors = Sensors()
        self.actuators = Actuators()

        # IMPORTANT:
        # There is NO internal state here.

    def decide(self, percept):

        available = (
            percept["available_spaces"]
        )

        waiting = (
            percept["cars_waiting"]
        )

        reserved = (
            percept["reserved_available"]
        )

        # ----------------------------------------------------
        # RULE 1: PARKING FULL
        # ----------------------------------------------------

        if available <= 0:

            return "NO_PARKING"

        # ----------------------------------------------------
        # RULE 2: CARS WAITING + SPACE AVAILABLE
        # ----------------------------------------------------

        elif waiting > 0 and available > 0:

            return "ALLOW_PARKING"

        # ----------------------------------------------------
        # RULE 3: RESERVED SLOT AVAILABLE
        # ----------------------------------------------------

        elif reserved and available > 0:

            return "ALLOW_PARKING"

        # ----------------------------------------------------
        # DEFAULT
        # ----------------------------------------------------

        else:

            return "WAIT"

    def run(self, environment):

        # Sensor reads environment
        percept = self.sensors.read(
            environment
        )

        # Agent makes decision
        action = self.decide(
            percept
        )

        # Actuator modifies environment
        result = self.actuators.execute(
            action,
            environment
        )

        return action, result


# ============================================================
# MODEL-BASED AGENT
# ============================================================

class ModelBasedAgent:

    def __init__(self):

        self.sensors = Sensors()
        self.actuators = Actuators()

        # ====================================================
        # INTERNAL STATE
        # ====================================================

        self.state = {

            "previous_spaces": None,

            "previous_waiting": None,

            "availability_trend": None,

            "previous_action": "NONE"
        }

    def run(self, environment):

        # ====================================================
        # SENSOR
        # ====================================================

        percept = self.sensors.read(
            environment
        )

        current_spaces = (
            percept["available_spaces"]
        )

        current_waiting = (
            percept["cars_waiting"]
        )

        # ====================================================
        # READ INTERNAL STATE
        # ====================================================

        previous_spaces = (
            self.state[
                "previous_spaces"
            ]
        )

        previous_action = (
            self.state[
                "previous_action"
            ]
        )

        # ====================================================
        # CALCULATE AVAILABILITY TREND
        # ====================================================

        if previous_spaces is None:

            trend = None

        else:

            trend = (
                current_spaces
                - previous_spaces
            )

        # ====================================================
        # MAKE DECISION
        # ====================================================

        if current_spaces <= 0:

            action = "NO_PARKING"

        elif current_waiting > 0:

            action = "ALLOW_PARKING"

        else:

            action = "WAIT"

        # ====================================================
        # ACTUATOR
        # ====================================================

        result = self.actuators.execute(
            action,
            environment
        )

        # ====================================================
        # UPDATE INTERNAL STATE
        # ====================================================

        self.state[
            "previous_spaces"
        ] = current_spaces

        self.state[
            "previous_waiting"
        ] = current_waiting

        self.state[
            "availability_trend"
        ] = trend

        self.state[
            "previous_action"
        ] = action

        # ====================================================
        # RETURN DATA
        # ====================================================

        return {

            "action":
                action,

            "result":
                result,

            "previous_spaces":
                previous_spaces,

            "availability_trend":
                trend,

            "previous_action":
                previous_action
        }


# ============================================================
# MAIN PROGRAM
# ============================================================

# ------------------------------------------------------------
# 1. CREATE ENVIRONMENT
# ------------------------------------------------------------
# This immediately creates the GUI.

environment = Environment()


# ------------------------------------------------------------
# 2. CREATE SIMPLE REFLEX AGENT
# ------------------------------------------------------------

simple_agent = SimpleReflexAgent()


# ------------------------------------------------------------
# 3. CREATE MODEL-BASED AGENT
# ------------------------------------------------------------

model_agent = ModelBasedAgent()


# ------------------------------------------------------------
# 4. CONNECT AGENTS TO ENVIRONMENT
# ------------------------------------------------------------

environment.simple_agent = simple_agent

environment.model_agent = model_agent


# ------------------------------------------------------------
# 5. START GUI
# ------------------------------------------------------------

environment.root.mainloop()
```

```python
class SimpleReflexAgent:

    def __init__(self):

        self.sensors = Sensors()
        self.actuators = Actuators()

    def decide(self, percept):

        available = (
            percept["available_spaces"]
        )

        waiting = (
            percept["cars_waiting"]
        )

        reserved = (
            percept["reserved_available"]
        )

        # =================================
        # RULE 1
        # =================================

        if available <= 0:

            return "NO_PARKING"

        # =================================
        # RULE 2
        # =================================

        elif waiting > 0 and available > 0:

            return "ALLOW_PARKING"

        # =================================
        # RULE 3
        # =================================

        elif reserved and available > 0:

            return "ALLOW_PARKING"

        # =================================
        # DEFAULT
        # =================================

        else:

            return "WAIT"

    def run(self, environment):

        # Sensor reads environment
        percept = self.sensors.read(
            environment
        )

        # Agent makes decision
        action = self.decide(percept)

        # Actuator changes environment
        result = self.actuators.execute(
            action,
            environment
        )

        return action, result
```

```python
class ModelBasedAgent:

    def __init__(self):

        self.sensors = Sensors()
        self.actuators = Actuators()

        # =================================
        # INTERNAL STATE
        # =================================

        self.state = {

            "previous_spaces": None,

            "previous_waiting": None,

            "availability_trend": None,

            "previous_action": "NONE"
        }

    def run(self, environment):

        # =================================
        # SENSOR
        # =================================

        percept = self.sensors.read(
            environment
        )

        current_spaces = (
            percept["available_spaces"]
        )

        current_waiting = (
            percept["cars_waiting"]
        )

        # =================================
        # READ OLD INTERNAL STATE
        # =================================

        previous_spaces = (
            self.state["previous_spaces"]
        )

        previous_action = (
            self.state["previous_action"]
        )

        # =================================
        # CALCULATE TREND
        # =================================

        if previous_spaces is None:

            trend = None

        else:

            trend = (
                current_spaces
                - previous_spaces
            )

        # =================================
        # MAKE DECISION
        # =================================

        if current_spaces <= 0:

            action = "NO_PARKING"

        elif current_waiting > 0:

            action = "ALLOW_PARKING"

        else:

            action = "WAIT"

        # =================================
        # ACTUATOR
        # =================================

        result = self.actuators.execute(
            action,
            environment
        )

        # =================================
        # UPDATE INTERNAL STATE
        # =================================

        self.state[
            "previous_spaces"
        ] = current_spaces

        self.state[
            "previous_waiting"
        ] = current_waiting

        self.state[
            "availability_trend"
        ] = trend

        self.state[
            "previous_action"
        ] = action

        # =================================
        # RETURN DATA
        # =================================

        return {

            "action": action,

            "result": result,

            "previous_spaces":
                previous_spaces,

            "availability_trend":
                trend,

            "previous_action":
                previous_action
        }
```
