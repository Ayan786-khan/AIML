## ERROR CALCULATION FOR LINEAR

```python
# Manual Evaluation of Linear Regression

# Number of observations
n = len(y_test)

# -----------------------------
# 1. Mean Squared Error (MSE)
# -----------------------------

mse_linear_manual = np.sum(
    (y_test - y_pred_linear) ** 2
) / n


# -----------------------------
# 2. Mean Absolute Error (MAE)
# -----------------------------

mae_linear_manual = np.sum(
    np.abs(y_test - y_pred_linear)
) / n


# -----------------------------
# 3. R2 Score
# -----------------------------

y_mean = np.mean(y_test)

ss_res = np.sum(
    (y_test - y_pred_linear) ** 2
)

ss_tot = np.sum(
    (y_test - y_mean) ** 2
)

r2_linear_manual = 1 - (ss_res / ss_tot)


# -----------------------------
# Print Results
# -----------------------------

print("*" * 40)
print("LINEAR REGRESSION EVALUATION RESULTS")
print("*" * 40)

results_linear = pd.DataFrame({
    "Evaluation Metric": [
        "Mean Squared Error (MSE)",
        "Mean Absolute Error (MAE)",
        "R2 Score"
    ],
    "Value": [
        mse_linear_manual,
        mae_linear_manual,
        r2_linear_manual
    ]
})

display(results_linear.round(4))

print("*" * 40)
```
