```python
import time
```

```python
maze = [
    "S.....",
    ".####.",
    "......",
    ".####.",
    "E.....",
]

for r in range(len(maze)):
    for c in range(len(maze[0])):
        if maze[r][c] == "S":
            start = (r, c)
        if maze[r][c] == "E":
            end = (r, c)
print("Start :", start, "End :", end)
```

```python
def get_neighbour(pos):
    r, c = pos
    result = []
    for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nr = dr + r
        nc = dc + c
        if 0 <= nr < len(maze) and 0 <= nc < len(maze[0]):
            if maze[nr][nc] != "#":
                result.append((nr, nc))
    return result
```

```python
def bfs(start, end):
    queue = [[start]]
    visited = [start]
    explored = 0
    while len(queue) > 0:
        path = queue.pop(0)
        node = path[-1]
        explored = explored + 1
        if node == end:
            return path, explored
        for n in get_neighbour(node):
            if n not in visited:
                visited.append(n)
                queue.append(path + [n])
    return None, explored
```

```python
def dfs(start, end):
    stack = [[start]]
    visited = [start]
    explored = 0
    while len(stack) > 0:
        path = stack.pop()
        node = path[-1]
        explored = explored + 1
        if node == end:
            return path, explored

        for n in get_neighbour(node):
            if n not in visited:
                visited.append(n)
                stack.append(path + [n])
    return None, explored
```

```python
t1 = time.perf_counter()
bfs_path, bfs_node = bfs(start, end)
bfs_time = (time.perf_counter() - t1) * 1000

t2 = time.perf_counter()
dfs_path, dfs_node = dfs(start, end)
dfs_time = (time.perf_counter() - t2) * 1000

print("Algorithm | path length | Nodes explored | Time(ms)")
print("BFS       |", len(bfs_path), "          |", bfs_node, "             |", round(bfs_time, 4))
print("DFS       |", len(dfs_path), "         |", dfs_node, "            |", round(dfs_time, 4))
```

```python
def show_path(path):
    grid = [list(row) for row in maze]
    for (r, c) in path:
        if grid[r][c] == ".":
            grid[r][c] = "*"
    for row in grid:
        print("".join(row))

print("BFS Path :")
show_path(bfs_path)
print("DFS path :")
show_path(dfs_path)
```
