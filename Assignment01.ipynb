## DATA CLEANING AND HANDLING MISSING VALUES

```python
print("*" * 70)
print("Handling Missing Values by Replacing with Column Median")
print("*" * 70)

df['bill_length_mm'] = df['bill_length_mm'].fillna(
    df['bill_length_mm'].median()
)

df['bill_depth_mm'] = df['bill_depth_mm'].fillna(
    df['bill_depth_mm'].median()
)

df['flipper_length_mm'] = df['flipper_length_mm'].fillna(
    df['flipper_length_mm'].median()
)

df['body_mass_g'] = df['body_mass_g'].fillna(
    df['body_mass_g'].median()
)

print(df.isnull().sum())
```

## ENCODING

```python
# Create encoder
encoder = LabelEncoder()

# Encode Male/Female
df['sex'] = encoder.fit_transform(df['sex'])
```

```python
df.drop_duplicates(inplace=True)
```

```python
encoder = LabelEncoder()

for col in df.select_dtypes(include='object').columns:
    df[col] = encoder.fit_transform(df[col])
```

## FEATURES SCALING

```python
# Select numerical features
X = df[['bill_length_mm',
        'bill_depth_mm',
        'flipper_length_mm',
        'body_mass_g']]

# Create scaler
scaler = StandardScaler()

# Scale the features
X_scaled = scaler.fit_transform(X)

# Convert scaled data back to DataFrame
X = pd.DataFrame(X_scaled, columns=X.columns)

print(X.head())
```
