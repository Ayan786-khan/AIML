```python
# ============================================================
# BLOCK 2: FORMAT LOGICAL EXPRESSIONS
# ============================================================

def format_expression(expr):

    if isinstance(expr, str):
        return expr

    operator = expr[0]
    operands = expr[1:]

    formatted_operands = []

    for operand in operands:
        formatted_operands.append(
            format_expression(operand)
        )

    return "(" + f" {operator} ".join(
        formatted_operands
    ) + ")"
```

```python
# ============================================================
# BLOCK 3: EVALUATE LOGICAL EXPRESSIONS
# ============================================================

def evaluate_expression(expr, facts):

    # Simple fact
    if isinstance(expr, str):
        return expr in facts

    operator = expr[0]
    operands = expr[1:]

    # AND
    if operator == "AND":

        return all(
            evaluate_expression(
                operand,
                facts
            )
            for operand in operands
        )

    # OR
    elif operator == "OR":

        return any(
            evaluate_expression(
                operand,
                facts
            )
            for operand in operands
        )

    return False
```

```python
# ============================================================
# BLOCK 4: DISPLAY FACTS, RULES AND GOALS
# ============================================================

def display_facts_and_rules(domain):

    print("\n" + "=" * 70)
    print(f"              {domain.upper()}")
    print("=" * 70)

    print("\nFACTS")
    print("-" * 70)

    facts = knowledge_base[domain]["facts"]

    for i, fact in enumerate(facts, 1):
        print(f"{i:2}. {fact}")

    print("\nRULES")
    print("-" * 70)

    rules = knowledge_base[domain]["rules"]

    for i, rule in enumerate(rules, 1):

        condition = format_expression(
            rule["if"]
        )

        conclusion = rule["then"]

        print(
            f"{i:2}. IF {condition} "
            f"THEN {conclusion}"
        )

    print("\nPOSSIBLE GOALS")
    print("-" * 70)

    goals = knowledge_base[domain]["goals"]

    for i, goal in enumerate(goals, 1):
        print(f"{i:2}. {goal}")
```

```python
# ============================================================
# BLOCK 5: GET INITIAL FACTS FOR FORWARD CHAINING
# ============================================================

def get_forward_facts(domain):

    facts = knowledge_base[domain]["facts"]

    print("\n" + "=" * 70)
    print("             FORWARD CHAINING")
    print("=" * 70)

    print("\nAvailable Facts:")
    print("-" * 70)

    for i, fact in enumerate(facts, 1):
        print(f"{i:2}. {fact}")

    print("\nEnter fact numbers separated by spaces.")
    print("Example: 1 2 3")

    while True:

        value = input(
            "\nEnter initial facts: "
        ).strip()

        try:

            numbers = [
                int(x)
                for x in value.split()
            ]

            if not numbers:
                print("Enter at least one fact.")
                continue

            if any(
                n < 1 or n > len(facts)
                for n in numbers
            ):
                print("Invalid fact number.")
                continue

            return [
                facts[n - 1]
                for n in numbers
            ]

        except ValueError:

            print(
                "Please enter numbers only."
            )
```

```python
# ============================================================
# BLOCK 6: FORWARD CHAINING
# ============================================================

def forward_chaining(
    domain,
    initial_facts
):

    rules = knowledge_base[domain]["rules"]

    facts = set(initial_facts)

    fired_rules = []

    changed = True

    while changed:

        changed = False

        for rule_number, rule in enumerate(
            rules,
            1
        ):

            condition = rule["if"]
            conclusion = rule["then"]

            if evaluate_expression(
                condition,
                facts
            ):

                if conclusion not in facts:

                    facts.add(conclusion)

                    fired_rules.append({
                        "rule_number": rule_number,
                        "condition": condition,
                        "conclusion": conclusion
                    })

                    changed = True

    return facts, fired_rules
```

```python
# ============================================================
# BLOCK 7: RUN FORWARD CHAINING
# ============================================================

def run_forward_chaining(domain):

    initial_facts = get_forward_facts(
        domain
    )

    final_facts, fired_rules = (
        forward_chaining(
            domain,
            initial_facts
        )
    )

    print("\n" + "=" * 70)
    print("             FORWARD CHAINING FLOW")
    print("=" * 70)

    print("\nSTART")
    print("  |")
    print("  v")

    print("\nINITIAL FACTS:")

    for fact in initial_facts:
        print(f"  -> {fact}")

    print("\n  |")
    print("  v")

    if fired_rules:

        for step, item in enumerate(
            fired_rules,
            1
        ):

            rule_number = item[
                "rule_number"
            ]

            condition = item[
                "condition"
            ]

            conclusion = item[
                "conclusion"
            ]

            print(f"\nSTEP {step}")

            print(
                f"Rule {rule_number}: "
                f"IF {format_expression(condition)}"
            )

            print(
                f"THEN {conclusion}"
            )

            print(
                f"  => Derived: {conclusion}"
            )

            print("  |")
            print("  v")

    else:

        print(
            "No rules were fired."
        )

    print("\nFINAL KNOWLEDGE:")

    for fact in sorted(final_facts):
        print(f"  -> {fact}")

    print("\n" + "=" * 70)
    print("          FORWARD CHAIN COMPLETED")
    print("=" * 70)
```

```python
# ============================================================
# BLOCK 8: GET GOAL FOR BACKWARD CHAINING
# ============================================================

def get_backward_goal(domain):

    goals = knowledge_base[domain]["goals"]

    print("\n" + "=" * 70)
    print("             BACKWARD CHAINING")
    print("=" * 70)

    print("\nAvailable Goals:")
    print("-" * 70)

    for i, goal in enumerate(
        goals,
        1
    ):
        print(f"{i:2}. {goal}")

    while True:

        value = input(
            "\nEnter goal number: "
        ).strip()

        try:

            number = int(value)

            if 1 <= number <= len(goals):

                return goals[
                    number - 1
                ]

            print("Invalid goal number.")

        except ValueError:

            print(
                "Please enter a valid number."
            )
```

```python
# ============================================================
# BLOCK 9: BACKWARD CHAINING
# ============================================================

def backward_chaining(
    domain,
    goal
):

    rules = knowledge_base[domain]["rules"]

    # Original facts are the base/supporting facts
    original_facts = set(
        knowledge_base[domain]["facts"]
    )

    trace = []

    visited = set()

    # --------------------------------------------------------
    # Find rules that produce a particular goal
    # --------------------------------------------------------

    def find_rules_for_goal(
        current_goal
    ):

        matching_rules = []

        for rule_number, rule in enumerate(
            rules,
            1
        ):

            if rule["then"] == current_goal:

                matching_rules.append(
                    (
                        rule_number,
                        rule["if"]
                    )
                )

        return matching_rules

    # --------------------------------------------------------
    # Prove an expression
    # --------------------------------------------------------

    def prove_expression(
        expression,
        depth=0
    ):

        indent = "    " * depth

        # ----------------------------------------------------
        # SIMPLE FACT
        # ----------------------------------------------------

        if isinstance(
            expression,
            str
        ):

            if expression in original_facts:

                trace.append(
                    f"{indent}SUPPORTING FACT FOUND: "
                    f"{expression}"
                )

                return True

            trace.append(
                f"{indent}Need to derive: "
                f"{expression}"
            )

            return prove_goal(
                expression,
                depth
            )

        # ----------------------------------------------------
        # LOGICAL EXPRESSION
        # ----------------------------------------------------

        operator = expression[0]
        operands = expression[1:]

        # ----------------------------------------------------
        # AND
        # ----------------------------------------------------

        if operator == "AND":

            trace.append(
                f"{indent}Checking AND:"
            )

            for operand in operands:

                trace.append(
                    f"{indent}  Need: "
                    f"{format_expression(operand)}"
                )

                result = prove_expression(
                    operand,
                    depth + 1
                )

                if not result:

                    trace.append(
                        f"{indent}AND FAILED"
                    )

                    return False

            trace.append(
                f"{indent}AND SUCCESS"
            )

            return True

        # ----------------------------------------------------
        # OR
        # ----------------------------------------------------

        elif operator == "OR":

            trace.append(
                f"{indent}Checking OR:"
            )

            for operand in operands:

                trace.append(
                    f"{indent}  Trying: "
                    f"{format_expression(operand)}"
                )

                result = prove_expression(
                    operand,
                    depth + 1
                )

                if result:

                    trace.append(
                        f"{indent}OR SUCCESS"
                    )

                    return True

            trace.append(
                f"{indent}OR FAILED"
            )

            return False

        return False

    # --------------------------------------------------------
    # Prove a goal
    # --------------------------------------------------------

    def prove_goal(
        current_goal,
        depth=0
    ):

        indent = "    " * depth

        # If goal is already a fact
        if current_goal in original_facts:

            trace.append(
                f"{indent}SUPPORTING FACT FOUND: "
                f"{current_goal}"
            )

            return True

        # Prevent infinite recursion
        if current_goal in visited:

            trace.append(
                f"{indent}Already checked: "
                f"{current_goal}"
            )

            return False

        visited.add(current_goal)

        trace.append(
            f"{indent}GOAL: {current_goal}"
        )

        # Find rules producing goal
        matching_rules = (
            find_rules_for_goal(
                current_goal
            )
        )

        if not matching_rules:

            trace.append(
                f"{indent}NO RULE FOUND FOR: "
                f"{current_goal}"
            )

            return False

        # Try each rule
        for rule_number, condition in matching_rules:

            condition_text = (
                format_expression(
                    condition
                )
            )

            trace.append(
                f"{indent}Trying Rule "
                f"{rule_number}:"
            )

            trace.append(
                f"{indent}  IF {condition_text}"
            )

            trace.append(
                f"{indent}  THEN {current_goal}"
            )

            trace.append(
                f"{indent}  Searching for "
                f"supporting conditions..."
            )

            result = prove_expression(
                condition,
                depth + 1
            )

            if result:

                trace.append(
                    f"{indent}RULE "
                    f"{rule_number} SUCCESS"
                )

                trace.append(
                    f"{indent}PROVED: "
                    f"{current_goal}"
                )

                return True

            else:

                trace.append(
                    f"{indent}RULE "
                    f"{rule_number} FAILED"
                )

        trace.append(
            f"{indent}FAILED TO PROVE: "
            f"{current_goal}"
        )

        return False

    # --------------------------------------------------------
    # Start from goal
    # --------------------------------------------------------

    result = prove_goal(
        goal
    )

    return result, trace
```

```python
# ============================================================
# BLOCK 10: RUN BACKWARD CHAINING
# ============================================================

def run_backward_chaining(domain):

    goal = get_backward_goal(
        domain
    )

    print("\n" + "=" * 70)
    print("            BACKWARD REASONING FLOW")
    print("=" * 70)

    print("\nSTARTING GOAL:")
    print(f"  {goal}")

    print("\n  |")
    print("  v")

    result, trace = backward_chaining(
        domain,
        goal
    )

    print("\nREASONING TRACE:")
    print("-" * 70)

    for line in trace:
        print(line)

    print("\n" + "=" * 70)
    print("                  FINAL RESULT")
    print("=" * 70)

    if result:

        print(
            f"\nGOAL PROVED: {goal}"
        )

        print(
            "\nThe goal was successfully "
            "connected to supporting facts."
        )

    else:

        print(
            f"\nGOAL NOT PROVED: {goal}"
        )

        print(
            "\nThe required supporting facts "
            "could not be established."
        )

    print("\n" + "=" * 70)
    print("          BACKWARD CHAIN COMPLETED")
    print("=" * 70)
```

```python
# ============================================================
# BLOCK 11: MAIN MENU
# ============================================================

def main():

    while True:

        print("\n")
        print("=" * 70)
        print("                    EXPERT SYSTEM")
        print("=" * 70)

        print("\nSelect Expert System:")

        domains = list(
            knowledge_base.keys()
        )

        for i, domain in enumerate(
            domains,
            1
        ):

            print(
                f"{i}. {domain}"
            )

        print("0. Exit")

        choice = input(
            "\nEnter your choice: "
        ).strip()

        # ----------------------------------------------------
        # EXIT
        # ----------------------------------------------------

        if choice == "0":

            print(
                "\nExiting Expert System..."
            )

            break

        # ----------------------------------------------------
        # VALIDATE
        # ----------------------------------------------------

        if not choice.isdigit():

            print(
                "Please enter a valid number."
            )

            continue

        choice = int(choice)

        if choice < 1 or choice > len(domains):

            print(
                "Invalid choice."
            )

            continue

        domain = domains[
            choice - 1
        ]

        # ====================================================
        # DOMAIN MENU
        # ====================================================

        while True:

            print("\n")
            print("=" * 70)
            print(
                f"Selected Domain: {domain}"
            )
            print("=" * 70)

            print(
                "1. Display Facts & Rules"
            )

            print(
                "2. Forward Chaining"
            )

            print(
                "3. Backward Chaining"
            )

            print(
                "4. Change Domain"
            )

            print(
                "0. Exit"
            )

            sub_choice = input(
                "\nEnter your choice: "
            ).strip()

            # ------------------------------------------------
            # DISPLAY
            # ------------------------------------------------

            if sub_choice == "1":

                display_facts_and_rules(
                    domain
                )

            # ------------------------------------------------
            # FORWARD
            # ------------------------------------------------

            elif sub_choice == "2":

                run_forward_chaining(
                    domain
                )

            # ------------------------------------------------
            # BACKWARD
            # ------------------------------------------------

            elif sub_choice == "3":

                run_backward_chaining(
                    domain
                )

            # ------------------------------------------------
            # CHANGE DOMAIN
            # ------------------------------------------------

            elif sub_choice == "4":

                break

            # ------------------------------------------------
            # EXIT
            # ------------------------------------------------

            elif sub_choice == "0":

                return

            else:

                print(
                    "Invalid choice."
                )


# ============================================================
# BLOCK 12: RUN PROGRAM
# ============================================================

main()
```
